**Task 4: Analysis Results**

- There are no missing values in any column.
- There is one duplicate row beyond its first occurrence.
- There are three species: setosa, versicolor, and virginica, with 50 flowers each.
- The average sepal length is approximately 5.843 cm.
- The maximum petal length is 6.9 cm.
- The minimum sepal width is 2.0 cm.
- Grouping by species shows that virginica has the highest average petal length at 5.552 cm, while setosa has the lowest at 1.462 cm.
- Descriptive statistics summarize the counts, means, standard deviations, minimums, quartiles, and maximums of the numeric columns.

```python
filtered_iris = df[df["sepal_length"] > 6]
filtered_iris
```

**Task 5: Filtering Results**

The dataset was filtered by sepal length greater than 6 cm, petal width greater than 1.5 cm, and the setosa species.

Virginica has the highest average petal length at 5.552 cm.

The data was sorted by sepal length in descending order.
